## Ordinal classifier implementation
No ordinal-regression package was available offline, so `OrdinalLogisticClassifier` implements the Frank & Hall (2001) binary decomposition directly: K-1 binary logistic classifiers, each predicting P(grade > k), recombined into class probabilities that respect the natural ordering of the grade scale. See `src/models/ordinal_classifier.py` for the full implementation and citation.

## Key results

1. **course_id dominates**: every model improves substantially when course_id is included (macro-F1 +0.07 to +0.12). Confirms the Stage 1 EDA finding quantitatively.
2. **Nominal vs ordinal tie on macro-F1/accuracy, but ordinal wins on MAE**: under `with_course`, ordinal LR has the lowest MAE (1.379) of any configuration — when it's wrong, it's wrong by less. Nominal classification has no mechanism to prefer 'close' errors. This is the real evidence for the earlier ordinal-vs-nominal question.
3. All baselines cluster closely within a feature-set condition — sets the bar Stage 4 (Random Forest / Gradient Boosting) needs to clear.
